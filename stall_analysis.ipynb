#**Introduction**

# This project analyzes 14 days of sales and stock data for Wanjiku's produce stall at Gikomba Market. The objective is to understand sales performance, identify strong and weak sales days, compare weekday and weekend performance, monitor stock levels, and calculate gross profit margins.

# The analysis uses core Python programming to process and analyse the sales data.

# Dataset

```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700] # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg

stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions": {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma": {"qty": 9, "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado": {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

# **Task 1: Daily Sales Analysis**

# Compute total, average, highest and lowest daily sales for the full 14 days, and identify the day name and week number of the best and worst days. (Hint: list.index() in Section 1.10 gives the position; positions 0–6 are week 1.)

```python
total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)

highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

highest_day = days[daily_sales.index(highest_sales)]
lowest_day = days[daily_sales.index(lowest_sales)]

print(f"Total sales: KES {total_sales:,.2f}")
print(f"Average daily sales: KES {average_sales:,.2f}")
print(f"Highest sales: KES {highest_sales:,.2f} ({highest_day})")
print(f"Lowest sales: KES {lowest_sales:,.2f} ({lowest_day})")
```

# **Task 2: Weekly Sales Comparison**
# Write a function week_totals(sales) that returns a tuple of (week 1 total, week 2 total) using slicing, then calculate the week-on-week growth percentage. (Hint: returning and unpacking two values is covered in Section 1.10.)

```python
def week_totals(sales):
    """Return total sales for week one and week two."""
    week_one = sum(sales[:7])
    week_two = sum(sales[7:14])
    return week_one, week_two

week_one, week_two = week_totals(daily_sales)
growth = ((week_two - week_one) / week_one) * 100

print(f"Week 1 sales: KES {week_one:,.2f}")
print(f"Week 2 sales: KES {week_two:,.2f}")
print(f"Week-on-week growth: {growth:.2f}%")
```

# **Task 3: Daily Performance Classification**
# Write a function classify_day(amount, average) that returns “Strong” (more than 20% above average), “Normal” or “Weak” (more than 20% below average). Apply it to every day and count how many days fall into each category using the dictionary counting pattern from Section 1.10.

```python
def classify_day(amount, average):
    """Classify daily sales as Strong, Normal, or Weak."""
    if amount > average * 1.20:
        return "Strong"
    elif amount < average * 0.80:
        return "Weak"
    else:
        return "Normal"
categories = {
    "Strong": 0,
    "Normal": 0,
    "Weak": 0
}
for amount in daily_sales:
    result = classify_day(amount, average_sales)
    categories[result] += 1

for category, count in categories.items():
    print(f"{category}: {count} days")
```

# **Task 4: Weekday Versus Weekend Analysis**
# Calculate the average sales for weekdays (Mon–Fri) versus weekends (Sat–Sun) and state the difference as a percentage.

```python
weekday_sales = daily_sales[0:5] + daily_sales[7:12]
weekend_sales = daily_sales[5:7] + daily_sales[12:14]

weekday_average = sum(weekday_sales) / len(weekday_sales)
weekend_average = sum(weekend_sales) / len(weekend_sales)

percentage_difference = (
    (weekday_average - weekend_average) / weekend_average
) * 100
print(f"Weekday average: KES {weekday_average:,.2f}")
print(f"Weekend average: KES {weekend_average:,.2f}")
print(f"Percentage difference: {percentage_difference:.2f}%")
```

# **Task 5: Stock Reorder Analysis/Gross Margin Percentage
# Using the stock dictionary, write a function that returns a list of products whose quantity is at or below the reorder level, and calculate the gross margin percentage, (price − cost) ÷ price, for each product. (Hint: loop with .items().)

```python
def get_reorder_list(stock_dict):
    """Return products that need replenishment and required quantities."""
    reorder_items = []
    for name, details in stock_dict.items():
        if details["qty"] <= details["reorder"]:
            quantity_needed = details["reorder"] - details["qty"]
            reorder_items.append({
                "name": name,
                "quantity_needed": quantity_needed
            })
    return reorder_items

reorder_list = get_reorder_list(stock)

for item in reorder_list:
    print(f'{item["name"]}: Reorder {item["quantity_needed"]} kg')

if not reorder_list:
    print("No products need reordering.")

# Calculate the gross margin percentage: ((price - cost) / price) * 100
def calculate_gross_margin(cost_price, selling_price):
    """Calculate gross profit margin as a percentage."""
    if selling_price <= 0:
        return 0
    gross_profit = selling_price - cost_price
    return (gross_profit / selling_price) * 100

print("\nGross Margins:")
for name, details in stock.items():
    margin = calculate_gross_margin(details["cost"], details["price"])
    print(f'{name}: {margin:.2f}%')
```

# **Task 6: Stall Performance Report
# Print a formatted “Stall Performance Report” using f-strings, with money formatted as KES with thousands separators.    

```python
print("=" * 50)
print("         STALL PERFORMANCE REPORT")
print("=" * 50)

print("\nSALES SUMMARY")
print(f"Total Sales: KES {total_sales:,.2f}")
print(f"Average Sales: KES {average_sales:,.2f}")
print(f"Highest Sales: KES {highest_sales:,.2f}")
print(f"Lowest Sales: KES {lowest_sales:,.2f}")

print("\nWEEKLY PERFORMANCE")
print(f"Week 1: KES {week_one:,.2f}")
print(f"Week 2: KES {week_two:,.2f}")
print(f"Growth: {growth:.2f}%")

print("\nDAILY CLASSIFICATION")
for category, count in categories.items():
    print(f"{category}: {count} days")

print("\nWEEKDAY VS WEEKEND")
print(f"Weekday Average: KES {weekday_average:,.2f}")
print(f"Weekend Average: KES {weekend_average:,.2f}")
print(f"Difference: {percentage_difference:.2f}%")

print("\nSTOCK REORDER")
for item in reorder_list:
    print(
        f'{item["name"]}: {item["quantity_needed"]} kg'
    )

print("\nGROSS PROFIT MARGINS")
for name, details in stock.items():
    margin = calculate_gross_margin(details["cost"], details["price"])
    print(f"{name}: {margin:.2f}%")

print("=" * 50)
```
